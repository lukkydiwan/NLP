```python
import pandas as pd
data = pd.read_csv('data.csv')
```

```python
data.head()
```

```python
data=data[['reviewText' , 'rating']]
```

```python
data.head()
```

```python
data.shape
```

```python
data.isnull().sum()
```

```python
#postive review is 1 and negative is 0


data['rating']=data['rating'].apply(lambda x: 0 if x<3 else 1)
```

```python
## lower all cases
data['reviewText']=data['reviewText'].str.lower()
```

```python
data.head()
```

```python
## removing special characters
import re
import nltk
from nltk.corpus import stopwords
from nltk import wordnet
```

```python
data['reviewText']=data['reviewText'].apply(lambda x: re.sub('[^a-z A-Z 0-9]+' , '' , x))
```

```python
data['reviewText']=data['reviewText'].apply(lambda x : " ".join([y for y in x.split() if y not in stopwords.words('english')]))
```

```python
data.head()
```

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')

lemmatizer=WordNetLemmatizer()
```

```python
def lemmatize_words(text):
    return " ".join([lemmatizer.lemmatize(word) for word in text.split()])
```

```python
data['reviewText'] = data['reviewText'].apply(lambda x : lemmatize_words(x))
```

```python
data.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(data['reviewText'],data['rating'], test_size=.2)
```
